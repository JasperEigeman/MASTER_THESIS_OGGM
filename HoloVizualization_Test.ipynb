```python
# ============================================================
# 1. IMPORTS AND HOLOVIZ SETUP
# ============================================================

from pathlib import Path

import numpy as np
import pandas as pd

import holoviews as hv
import panel as pn

# Enable HoloViews and Panel
hv.extension("bokeh")
pn.extension()

print("HoloViz initialized.")
```

```python
# ============================================================
# 2. RGI7 REGION AND EXPERIMENT CONFIGURATION
# ============================================================

RGI_REGION = 11

RGI_REGION_NAMES = {
    1: "Alaska",
    2: "Western Canada and USA",
    3: "Arctic Canada North",
    4: "Arctic Canada South",
    5: "Greenland Periphery",
    6: "Iceland",
    7: "Svalbard",
    8: "Scandinavia",
    9: "Russian Arctic",
    10: "North Asia",
    11: "Central Europe",
    12: "Caucasus and Middle East",
    13: "Central Asia",
    14: "South Asia West",
    15: "South Asia East",
    16: "Low Latitudes",
    17: "Southern Andes",
    18: "New Zealand",
    19: "Antarctic and Subantarctic",
}

REGION_NAME = RGI_REGION_NAMES[RGI_REGION]

GCM = "CCSM4"

print(
    f"RGI7 Region {RGI_REGION:02d}: {REGION_NAME}"
)
print(
    f"GCM: {GCM}"
)
```

```python
# ============================================================
# 3. MODEL OUTPUT DIRECTORY
# ============================================================

WORKDIR = (
    Path.home()
    / "oggm_work"
    / f"vas_rgi70_region{RGI_REGION}_cru_1901_2019"
)

print("Using:")
print(WORKDIR)

if not WORKDIR.exists():
    raise FileNotFoundError(
        f"Model output directory not found:\n{WORKDIR}"
    )
```

```python
# ============================================================
# 4. LOAD GLACIER-LEVEL RESULTS
# ============================================================

glacier_data = {}

# ------------------------------------------------------------
# Historical reconstruction
# ------------------------------------------------------------

glacier_data["VAS reconstruction"] = pd.read_csv(
    WORKDIR /
    "glacier_results_hist1901_2019.csv"
)


# ------------------------------------------------------------
# Future projections
# ------------------------------------------------------------

glacier_data["RCP2.6"] = pd.read_csv(
    WORKDIR /
    f"glacier_results_future_{GCM}_rcp26.csv"
)

glacier_data["RCP4.5"] = pd.read_csv(
    WORKDIR /
    f"glacier_results_future_{GCM}_rcp45.csv"
)

glacier_data["RCP8.5"] = pd.read_csv(
    WORKDIR /
    f"glacier_results_future_{GCM}_rcp85.csv"
)


# ------------------------------------------------------------
# Standardize year column
# ------------------------------------------------------------

for scenario, df in glacier_data.items():

    df["year"] = (
        df["year"]
        .astype(float)
        .astype(int)
    )

    glacier_data[scenario] = df


# ------------------------------------------------------------
# Basic checks
# ------------------------------------------------------------

for scenario, df in glacier_data.items():

    print(
        f"{scenario}: "
        f"{df['rgi_id'].nunique()} glaciers, "
        f"{df['year'].min()}-{df['year'].max()}"
    )
```

```python
# ============================================================
# 5. LOAD REGIONAL RESULTS
# ============================================================

regional_data = {}

regional_data["VAS reconstruction"] = pd.read_csv(
    WORKDIR /
    "regional_results_hist1901_2019.csv"
)

regional_data["RCP2.6"] = pd.read_csv(
    WORKDIR /
    f"regional_results_future_{GCM}_rcp26.csv"
)

regional_data["RCP4.5"] = pd.read_csv(
    WORKDIR /
    f"regional_results_future_{GCM}_rcp45.csv"
)

regional_data["RCP8.5"] = pd.read_csv(
    WORKDIR /
    f"regional_results_future_{GCM}_rcp85.csv"
)


for scenario, df in regional_data.items():

    df["year"] = (
        df["year"]
        .astype(float)
        .astype(int)
    )

    regional_data[scenario] = df


print("Regional results loaded.")
```

```python
# ============================================================
# 6. COLOURS AND PLOT SETTINGS
# ============================================================

COLORS = {
    "VAS reconstruction": "#DAA520",  # goldenrod
    "RCP2.6": "#0000FF",
    "RCP4.5": "#79BCFF",
    "RCP8.5": "#FF0000"
}


METRICS = {
    "Glacier volume": "volume_km3",
    "Glacier area": "area_km2",
    "Total glacier length": "length_km"
}


METRIC_LABELS = {
    "volume_km3": "Glacier volume (km³)",
    "area_km2": "Glacier area (km²)",
    "length_km": "Sum of glacier lengths (km)"
}


SCENARIOS = [
    "VAS reconstruction",
    "RCP2.6",
    "RCP4.5",
    "RCP8.5"
]
```

```python
# ============================================================
# 7. INTERACTIVE REGIONAL EVOLUTION
# ============================================================

metric_widget = pn.widgets.Select(
    name="Variable",
    options=METRICS,
    value="volume_km3"
)


scenario_widget = pn.widgets.Select(
    name="Scenario",
    options=[
        "All scenarios",
        "VAS reconstruction",
        "RCP2.6",
        "RCP4.5",
        "RCP8.5"
    ],
    value="All scenarios"
)


@pn.depends(
    metric_widget.param.value,
    scenario_widget.param.value
)
def regional_evolution(metric, selected_scenario):

    curves = []


    # --------------------------------------------------------
    # Determine which scenarios to plot
    # --------------------------------------------------------

    if selected_scenario == "All scenarios":

        scenarios_to_plot = SCENARIOS

    else:

        scenarios_to_plot = [
            selected_scenario
        ]


    # --------------------------------------------------------
    # Create curves
    # --------------------------------------------------------

    for scenario in scenarios_to_plot:

        df = regional_data[scenario]


        curves.append(
            hv.Curve(
                df,
                kdims="year",
                vdims=metric,
                label=scenario
            ).opts(
                color=COLORS[scenario],
                line_width=3,
                tools=["hover"]
            )
        )


    # --------------------------------------------------------
    # 2019 transition
    # --------------------------------------------------------

    transition = hv.VLine(
        2019
    ).opts(
        color="black",
        line_dash="dashed",
        line_width=1
    )


    return (
        hv.Overlay(curves)
        * transition
    ).opts(
        width=950,
        height=500,
        xlabel="Year",
        ylabel=METRIC_LABELS[metric],
        title=(
            f"RGI7 Region {RGI_REGION:02d} - "
            f"{REGION_NAME}"
        ),
        legend_position="top_left"
    )


regional_plot = pn.Column(
    pn.Row(
        metric_widget,
        scenario_widget
    ),
    regional_evolution
)


regional_plot
```

```python
# ============================================================
# 8. INDIVIDUAL GLACIER EXPLORER
# ============================================================

glacier_ids = sorted(
    glacier_data["VAS reconstruction"]["rgi_id"]
    .dropna()
    .unique()
)


glacier_widget = pn.widgets.Select(
    name="RGI7 glacier",
    options=glacier_ids,
    value=glacier_ids[0]
)


glacier_metric_widget = pn.widgets.Select(
    name="Variable",
    options=METRICS,
    value="volume_km3"
)


glacier_scenario_widget = pn.widgets.Select(
    name="Scenario",
    options=[
        "All scenarios",
        "VAS reconstruction",
        "RCP2.6",
        "RCP4.5",
        "RCP8.5"
    ],
    value="All scenarios"
)


@pn.depends(
    glacier_widget.param.value,
    glacier_metric_widget.param.value,
    glacier_scenario_widget.param.value
)
def glacier_evolution(
    glacier_id,
    metric,
    selected_scenario
):

    curves = []


    if selected_scenario == "All scenarios":

        scenarios_to_plot = SCENARIOS

    else:

        scenarios_to_plot = [
            selected_scenario
        ]


    for scenario in scenarios_to_plot:

        df = glacier_data[scenario]


        subset = df[
            df["rgi_id"] == glacier_id
        ]


        if subset.empty:
            continue


        curves.append(
            hv.Curve(
                subset,
                kdims="year",
                vdims=metric,
                label=scenario
            ).opts(
                color=COLORS[scenario],
                line_width=3,
                tools=["hover"]
            )
        )


    transition = hv.VLine(
        2019
    ).opts(
        color="black",
        line_dash="dashed",
        line_width=1
    )


    return (
        hv.Overlay(curves)
        * transition
    ).opts(
        width=950,
        height=500,
        xlabel="Year",
        ylabel=METRIC_LABELS[metric],
        title=glacier_id,
        legend_position="top_left"
    )


glacier_plot = pn.Column(
    pn.Row(
        glacier_widget,
        glacier_metric_widget,
        glacier_scenario_widget
    ),
    glacier_evolution
)


glacier_plot
```

```python
# ============================================================
# 9. INDIVIDUAL GLACIER RESPONSE
# ============================================================

def calculate_response(
    df,
    scenario
):

    start = df[
        df["year"] == 2019
    ][
        [
            "rgi_id",
            "volume_km3",
            "area_km2"
        ]
    ].copy()


    start = start.rename(
        columns={
            "volume_km3":
                "volume_2019_km3",

            "area_km2":
                "area_2019_km2"
        }
    )


    end = df[
        df["year"] == 2100
    ][
        [
            "rgi_id",
            "volume_km3"
        ]
    ].copy()


    end = end.rename(
        columns={
            "volume_km3":
                "volume_2100_km3"
        }
    )


    response = start.merge(
        end,
        on="rgi_id",
        how="inner"
    )


    response = response[
        response["volume_2019_km3"] > 0
    ].copy()


    response[
        "volume_loss_percent"
    ] = (
        (
            response["volume_2019_km3"]
            -
            response["volume_2100_km3"]
        )
        /
        response["volume_2019_km3"]
        *
        100
    )


    response["scenario"] = scenario


    return response


response_data = pd.concat(
    [
        calculate_response(
            glacier_data["RCP2.6"],
            "RCP2.6"
        ),

        calculate_response(
            glacier_data["RCP4.5"],
            "RCP4.5"
        ),

        calculate_response(
            glacier_data["RCP8.5"],
            "RCP8.5"
        )
    ],
    ignore_index=True
)


response_scenario_widget = pn.widgets.Select(
    name="Scenario",
    options=[
        "RCP2.6",
        "RCP4.5",
        "RCP8.5"
    ],
    value="RCP8.5"
)


@pn.depends(
    response_scenario_widget.param.value
)
def response_scatter(
    scenario
):

    df = response_data[
        response_data["scenario"] == scenario
    ]


    return hv.Scatter(
        df,
        kdims="volume_2019_km3",
        vdims=[
            "volume_loss_percent",
            "rgi_id",
            "area_2019_km2"
        ]
    ).opts(
        width=950,
        height=550,
        color=COLORS[scenario],
        size=8,
        tools=["hover"],
        xlabel="Glacier volume in 2019 (km³)",
        ylabel="Volume loss 2019–2100 (%)",
        title=(
            f"Individual glacier response - "
            f"{scenario}"
        )
    )


response_plot = pn.Column(
    response_scenario_widget,
    response_scatter
)


response_plot
```

```python
# ============================================================
# 10. NORMALIZED REGIONAL VOLUME
# ============================================================

normalized_data = {}


for scenario in SCENARIOS:

    df = regional_data[
        scenario
    ].copy()


    if scenario == "VAS reconstruction":

        reference = df.loc[
            df["year"] == 2019,
            "volume_km3"
        ].iloc[0]


    else:

        reference = df.loc[
            df["year"] == 2019,
            "volume_km3"
        ].iloc[0]


    df[
        "volume_percent_2019"
    ] = (
        df["volume_km3"]
        /
        reference
        *
        100
    )


    normalized_data[
        scenario
    ] = df


normalized_curves = []


for scenario in SCENARIOS:

    normalized_curves.append(
        hv.Curve(
            normalized_data[scenario],
            kdims="year",
            vdims="volume_percent_2019",
            label=scenario
        ).opts(
            color=COLORS[scenario],
            line_width=3,
            tools=["hover"]
        )
    )


normalized_plot = (
    hv.Overlay(normalized_curves)
    *
    hv.VLine(2019).opts(
        color="black",
        line_dash="dashed",
        line_width=1
    )
).opts(
    width=950,
    height=500,
    xlabel="Year",
    ylabel="Regional glacier volume (% of 2019)",
    title=(
        f"RGI7 Region {RGI_REGION:02d} - "
        f"{REGION_NAME}\n"
        "Normalized regional glacier volume"
    ),
    legend_position="top_left"
)


normalized_plot
```

```python
# ============================================================
# 11. SEA-LEVEL EQUIVALENT
# ============================================================

RHO_ICE = 900.0
RHO_WATER = 1000.0
OCEAN_AREA_KM2 = 3.625e8


volume_1901 = regional_data[
    "VAS reconstruction"
].loc[
    regional_data[
        "VAS reconstruction"
    ]["year"] == 1901,
    "volume_km3"
].iloc[0]


sle_data = {}


for scenario in SCENARIOS:

    df = regional_data[
        scenario
    ].copy()


    df[
        "volume_loss_from_1901_km3"
    ] = (
        volume_1901
        -
        df["volume_km3"]
    )


    df[
        "sea_level_change_mm"
    ] = (
        df[
            "volume_loss_from_1901_km3"
        ]
        *
        (RHO_ICE / RHO_WATER)
        /
        OCEAN_AREA_KM2
        *
        1e6
    )


    sle_data[
        scenario
    ] = df


sle_curves = []


for scenario in SCENARIOS:

    sle_curves.append(
        hv.Curve(
            sle_data[scenario],
            kdims="year",
            vdims="sea_level_change_mm",
            label=scenario
        ).opts(
            color=COLORS[scenario],
            line_width=3,
            tools=["hover"]
        )
    )


sle_plot = (
    hv.Overlay(sle_curves)
    *
    hv.VLine(2019).opts(
        color="black",
        line_dash="dashed",
        line_width=1
    )
).opts(
    width=950,
    height=500,
    xlabel="Year",
    ylabel="Sea-level equivalent relative to 1901 (mm)",
    title=(
        f"RGI7 Region {RGI_REGION:02d} - "
        f"{REGION_NAME}\n"
        "Glacier contribution to sea level"
    ),
    legend_position="top_left"
)


sle_plot
```

```python
# ============================================================
# 13. PREPARE GLACIER EVOLUTION DATA
# ============================================================

# ------------------------------------------------------------
# 1901 baseline for every glacier
# ------------------------------------------------------------

historical = glacier_data["VAS reconstruction"].copy()

baseline_1901 = (
    historical[
        historical["year"] == 1901
    ][
        [
            "rgi_id",
            "volume_km3",
            "area_km2",
            "length_km"
        ]
    ]
    .rename(
        columns={
            "volume_km3": "volume_1901_km3",
            "area_km2": "area_1901_km2",
            "length_km": "length_1901_km"
        }
    )
)


# ------------------------------------------------------------
# Create normalized datasets
# ------------------------------------------------------------

evolution_data = {}


for scenario, df in glacier_data.items():

    df = df.copy()

    df = df.merge(
        baseline_1901,
        on="rgi_id",
        how="inner"
    )

    # Exclude glaciers without a meaningful 1901 volume
    df = df[
        df["volume_1901_km3"] > 0
    ].copy()

    # Relative volume
    df["volume_pct_1901"] = (
        df["volume_km3"]
        / df["volume_1901_km3"]
        * 100
    )

    # Relative area
    df["area_pct_1901"] = (
        df["area_km2"]
        / df["area_1901_km2"]
        * 100
    )

    # Relative length
    df["length_pct_1901"] = (
        df["length_km"]
        / df["length_1901_km"]
        * 100
    )

    # Positive = loss relative to 1901
    df["volume_loss_pct"] = (
        100
        - df["volume_pct_1901"]
    )

    df["area_loss_pct"] = (
        100
        - df["area_pct_1901"]
    )

    evolution_data[scenario] = df


for scenario, df in evolution_data.items():

    print(
        scenario,
        ":",
        df["rgi_id"].nunique(),
        "glaciers,",
        df["year"].min(),
        "-",
        df["year"].max()
    )
```

```python
# ============================================================
# #13 Load RGI7 shapefile and prepare map geometry
# ============================================================

import glob
import os
import geopandas as gpd
import geoviews as gv
import cartopy.crs as ccrs
import numpy as np
import pandas as pd
import holoviews as hv
import panel as pn

# ------------------------------------------------------------
# Find the RGI7 regional shapefile
# ------------------------------------------------------------

RGI7_SHAPEFILE = None   # Set manually if automatic search fails

if RGI7_SHAPEFILE is None:

    search_roots = [
        os.path.expanduser("~/OGGM"),
        os.path.expanduser("~/.cache/oggm"),
        os.path.expanduser("~/oggm_work"),
    ]

    candidates = []

    for root in search_roots:
        for pattern in [
            "**/*.shp",
            "**/*.SHP",
        ]:
            candidates.extend(
                glob.glob(
                    os.path.join(root, pattern),
                    recursive=True
                )
            )

    region_tag = f"RGI2000-v7.0-G-{RGI_REGION:02d}-"

    for shp in candidates:

        try:
            gdf_test = gpd.read_file(shp)

            if "rgi_id" in gdf_test.columns:
                ids = gdf_test["rgi_id"].astype(str)
            elif "RGIId" in gdf_test.columns:
                ids = gdf_test["RGIId"].astype(str)
            else:
                continue

            if ids.astype(str).str.startswith(region_tag).any():
                RGI7_SHAPEFILE = shp
                break

        except Exception:
            continue


if RGI7_SHAPEFILE is None:
    raise FileNotFoundError(
        "Could not find the RGI7 shapefile automatically.\n"
        "Set RGI7_SHAPEFILE manually to the .shp file."
    )

print("Using RGI7 shapefile:")
print(RGI7_SHAPEFILE)


# ------------------------------------------------------------
# Read shapefile
# ------------------------------------------------------------

rgi7_map = gpd.read_file(RGI7_SHAPEFILE)

if "rgi_id" not in rgi7_map.columns:
    if "RGIId" in rgi7_map.columns:
        rgi7_map = rgi7_map.rename(columns={"RGIId": "rgi_id"})
    else:
        raise KeyError("No rgi_id or RGIId column found.")


rgi7_map["rgi_id"] = rgi7_map["rgi_id"].astype(str)

# Keep only glaciers which actually occur in the model outputs
model_ids = set(
    glacier_data["VAS reconstruction"]["rgi_id"].astype(str)
)

rgi7_map = rgi7_map[
    rgi7_map["rgi_id"].isin(model_ids)
].copy()

print(f"Glaciers available for map: {len(rgi7_map)}")


# ------------------------------------------------------------
# Keep geographic coordinates for selection
# ------------------------------------------------------------

if rgi7_map.crs is None:
    raise ValueError("RGI7 shapefile has no CRS defined.")

rgi7_map = rgi7_map.to_crs("EPSG:4326")

rep_points_ll = rgi7_map.geometry.representative_point()

rgi7_map["lon"] = rep_points_ll.x
rgi7_map["lat"] = rep_points_ll.y


# ------------------------------------------------------------
# Project map geometry to Web Mercator
# ------------------------------------------------------------

rgi7_map = rgi7_map.to_crs("EPSG:3857")

rep_points_merc = rgi7_map.geometry.representative_point()

rgi7_map["x_merc"] = rep_points_merc.x
rgi7_map["y_merc"] = rep_points_merc.y


MAP_CRS = ccrs.GOOGLE_MERCATOR


print("Map CRS:", rgi7_map.crs)
```

```python
# ============================================================
# #14 Interactive controls
# ============================================================

year_player = pn.widgets.Player(
    name="Year",
    start=1901,
    end=2100,
    value=2019,
    step=1,
    interval=250,
    loop_policy="loop",
)

scenario_animation_widget = pn.widgets.Select(
    name="Scenario",
    options=[
        "VAS reconstruction",
        "RCP2.6",
        "RCP4.5",
        "RCP8.5",
    ],
    value="VAS reconstruction",
)

print("Interactive controls ready.")
```

```python
# ============================================================
# #15 Dynamic glacier map + persistent click layer
# ============================================================

# ------------------------------------------------------------
# Persistent invisible click layer
# This is the SAME object used later by the Tap stream.
# ------------------------------------------------------------

click_points = gv.Points(
    rgi7_map,
    kdims=["x_merc", "y_merc"],
    vdims=[
        "rgi_id",
        "lon",
        "lat",
    ],
    crs=MAP_CRS,
).opts(
    size=14,
    fill_alpha=0.01,
    line_alpha=0,
    tools=["tap"],
)


def glacier_map(year, scenario):

    # --------------------------------------------------------
    # Select scenario
    # --------------------------------------------------------

    df = glacier_data[scenario].copy()

    df["rgi_id"] = df["rgi_id"].astype(str)
    df["year"] = pd.to_numeric(
        df["year"],
        errors="coerce"
    )

    # --------------------------------------------------------
    # Check expected columns
    # --------------------------------------------------------

    required_columns = [
        "year",
        "rgi_id",
        "volume_km3",
        "area_km2",
        "length_km",
    ]

    missing = [
        c for c in required_columns
        if c not in df.columns
    ]

    if missing:
        raise KeyError(
            f"Missing columns in {scenario}: {missing}"
        )

    # --------------------------------------------------------
    # Select year
    # --------------------------------------------------------

    df = df[
        df["year"] == year
    ].copy()

    if df.empty:
        return hv.Text(
            0,
            0,
            f"No model data available for {scenario}, {year}"
        ).opts(
            width=1000,
            height=650
        )

    # --------------------------------------------------------
    # Spatial data ONLY
    # Prevent duplicate area/other columns
    # --------------------------------------------------------

    spatial_df = rgi7_map[
        [
            "rgi_id",
            "geometry",
            "lon",
            "lat",
            "x_merc",
            "y_merc",
        ]
    ].copy()

    spatial_df["rgi_id"] = (
        spatial_df["rgi_id"].astype(str)
    )

    # --------------------------------------------------------
    # Merge model + spatial data
    # --------------------------------------------------------

    map_df = spatial_df.merge(
        df,
        on="rgi_id",
        how="inner"
    )

    if map_df.empty:
        return hv.Text(
            0,
            0,
            "No glaciers available for this year/scenario."
        ).opts(
            width=1000,
            height=650
        )

    # --------------------------------------------------------
    # 1901 baseline volume
    # --------------------------------------------------------

    hist_df = glacier_data[
        "VAS reconstruction"
    ].copy()

    hist_df["rgi_id"] = (
        hist_df["rgi_id"].astype(str)
    )

    baseline = (
        hist_df[
            hist_df["year"] == 1901
        ][
            [
                "rgi_id",
                "volume_km3",
            ]
        ]
        .rename(
            columns={
                "volume_km3":
                    "volume_1901_km3"
            }
        )
    )

    map_df = map_df.merge(
        baseline,
        on="rgi_id",
        how="left"
    )

    # --------------------------------------------------------
    # Volume relative to 1901
    # --------------------------------------------------------

    map_df["volume_pct_1901"] = (
        100
        * map_df["volume_km3"]
        / map_df["volume_1901_km3"]
    )

    map_df["volume_loss_pct"] = (
        100
        - map_df["volume_pct_1901"]
    )

    map_df["volume_pct_1901"] = (
        map_df["volume_pct_1901"]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .fillna(0)
    )

    map_df["volume_loss_pct"] = (
        map_df["volume_loss_pct"]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .fillna(0)
    )

    # --------------------------------------------------------
    # Polygon data
    # --------------------------------------------------------

    polygon_df = map_df[
        [
            "geometry",
            "rgi_id",
            "volume_km3",
            "area_km2",
            "length_km",
            "volume_pct_1901",
            "volume_loss_pct",
        ]
    ].copy()

    polygons = gv.Polygons(
        polygon_df,
        vdims=[
            "rgi_id",
            "volume_km3",
            "area_km2",
            "length_km",
            "volume_pct_1901",
            "volume_loss_pct",
        ],
        crs=MAP_CRS,
    ).opts(
        fill_color="lightgray",
        fill_alpha=0.15,
        line_color="gray",
        line_alpha=0.35,
        line_width=0.5,
        tools=["hover"],
    )

    # --------------------------------------------------------
    # Visible modeled glacier markers
    # --------------------------------------------------------

    points = gv.Points(
        map_df,
        kdims=[
            "x_merc",
            "y_merc",
        ],
        vdims=[
            "rgi_id",
            "lon",
            "lat",
            "volume_km3",
            "area_km2",
            "length_km",
            "volume_pct_1901",
            "volume_loss_pct",
        ],
        crs=MAP_CRS,
    ).opts(
        size=7,
        color="volume_loss_pct",
        cmap="Viridis",
        colorbar=True,
        tools=["hover"],
        alpha=0.9,
    )

    # --------------------------------------------------------
    # OSM
    # --------------------------------------------------------

    tiles = gv.tile_sources.OSM

    # --------------------------------------------------------
    # Final map
    # click_points MUST be included here
    # --------------------------------------------------------

    return (
        tiles
        * polygons
        * points
        * click_points
    ).opts(
        width=1000,
        height=650,
        active_tools=["wheel_zoom"],
        show_grid=False,
    )


# ------------------------------------------------------------
# Reactive map
# ------------------------------------------------------------

@pn.depends(
    year_player.param.value,
    scenario_animation_widget.param.value,
)
def map_view(year, scenario):

    return glacier_map(
        year=year,
        scenario=scenario
    )
```

```python
# ============================================================
# #16 Click glacier -> show full trajectory
# ============================================================

tap = hv.streams.Tap(
    source=click_points
)


@pn.depends(
    tap.param.x,
    tap.param.y
)
def selected_glacier_evolution(x, y):

    # --------------------------------------------------------
    # Nothing selected yet
    # --------------------------------------------------------

    if x is None or y is None:

        return pn.pane.Markdown(
            """
### Glacier trajectory

Click a glacier marker on the map to inspect its
1901–2100 volume evolution.
"""
        )


    # --------------------------------------------------------
    # Find nearest glacier to click
    # --------------------------------------------------------

    distances = (
        (rgi7_map["x_merc"] - x) ** 2
        +
        (rgi7_map["y_merc"] - y) ** 2
    )

    nearest_idx = distances.idxmin()

    glacier_id = str(
        rgi7_map.loc[
            nearest_idx,
            "rgi_id"
        ]
    )

    lon = float(
        rgi7_map.loc[
            nearest_idx,
            "lon"
        ]
    )

    lat = float(
        rgi7_map.loc[
            nearest_idx,
            "lat"
        ]
    )


    # --------------------------------------------------------
    # Scenario colours
    # --------------------------------------------------------

    scenario_colors = {
        "VAS reconstruction": "#DAA520",
        "RCP2.6": "#0000FF",
        "RCP4.5": "#79BCFF",
        "RCP8.5": "#FF0000",
    }


    # --------------------------------------------------------
    # Build curves
    # --------------------------------------------------------

    curves = []

    for scenario in [
        "VAS reconstruction",
        "RCP2.6",
        "RCP4.5",
        "RCP8.5",
    ]:

        if scenario not in glacier_data:
            continue

        d = glacier_data[
            scenario
        ].copy()

        d["rgi_id"] = (
            d["rgi_id"].astype(str)
        )

        d = d[
            d["rgi_id"] == glacier_id
        ].copy()

        if d.empty:
            continue

        d = d.sort_values("year")

        curve = hv.Curve(
            d[
                [
                    "year",
                    "volume_km3",
                ]
            ],
            kdims="year",
            vdims="volume_km3",
            label=scenario,
        ).opts(
            color=scenario_colors[scenario],
            line_width=3,
            tools=["hover"],
        )

        curves.append(curve)


    # --------------------------------------------------------
    # No trajectory found
    # --------------------------------------------------------

    if not curves:

        return pn.pane.Markdown(
            f"""
### {glacier_id}

Location: **{lon:.3f}°E, {lat:.3f}°N**

No model trajectory found.
"""
        )


    # --------------------------------------------------------
    # Combine curves
    # --------------------------------------------------------

    trajectory = hv.Overlay(
        curves
    ).opts(
        width=850,
        height=450,
        xlabel="Year",
        ylabel="Glacier volume (km³)",
        legend_position="top_left",
        show_grid=True,
    )

    transition = hv.VLine(
        2019
    ).opts(
        line_dash="dashed",
        line_width=2,
        color="black",
    )


    # --------------------------------------------------------
    # Selected glacier information
    # --------------------------------------------------------

    title = pn.pane.Markdown(
        f"""
### {glacier_id}

**Location:** {lon:.3f}°E, {lat:.3f}°N

**Model:** VAS historical reconstruction + CMIP5 RCP projections

The dashed line marks the transition from the historical
reconstruction to the future projection period.
"""
    )


    return pn.Column(
        title,
        trajectory * transition,
    )
```
